## Overview

This notebook integrates the processed demographic datasets into a unified representative population pool. The resulting population pool serves as the demographic foundation for the subsequent Behaviour Layer, Opportunity Layer, and mobility simulation framework.

Input datasets:

- Age
- Sex
- Household Composition
- Economic Activity Status
- Industry
- Car Ownership
- Distance Travelled to Work
- Ethnic Group

Output:

- Representative Population Pool

## 1. Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
```

## 2. Set Up File Pathes

```python
from pathlib import Path

project_root = Path.cwd().parents[1]

processed_demographic_path = (
    project_root
    / "data"
    / "demographic"
    / "processed"
)

processed_mobility_path = (
    project_root
    / "data"
    / "mobility"
    / "processed"
)

output_path = (
    project_root
    / "output"
    / "demographic_profiles"
)

output_path.mkdir(parents=True, exist_ok=True)
```

## 3. Load Dataset

```python
# Age
age_file = (
    processed_demographic_path
    / "age"
    / "age.json"
)

# Gender
gender_file = (
    processed_demographic_path
    / "gender"
    / "gender.json"
)

# Household Composition
household_file = (
    processed_demographic_path
    / "household_by_age"
    / "household_by_age.json"
)

# Economic Activity Status
economic_activity_file = (
    processed_demographic_path
    / "economic_activity_status"
    / "economic_activity_status.json"
)

# Industry
industry_file = (
    processed_demographic_path
    / "industry"
    / "industry_by_age.json"
)

# Car Ownership
car_ownership_file = (
    processed_demographic_path
    / "car_ownership"
    / "car_ownership.json"
)

# Distance Travelled to Work
travel_distance_file = (
    project_root
    / "data"
    / "mobility"
    / "processed"
    / "travel_distance"
    / "travel_distance_by_age.json"
)

# Ethnic Group
ethnic_group_file = (
    processed_demographic_path
    / "ethnic_group"
    / "ethnic_group_by_age.json"
)
```

```python
age = pd.read_json(age_file)

gender = pd.read_json(gender_file)

household = pd.read_json(household_file)

economic_activity = pd.read_json(economic_activity_file)

industry = pd.read_json(industry_file)

car_ownership = pd.read_json(car_ownership_file)

travel_distance = pd.read_json(travel_distance_file)

ethnic_group = pd.read_json(ethnic_group_file)
```
