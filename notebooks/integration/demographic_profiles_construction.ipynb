## Overview

This notebook integrates the processed demographic datasets into a unified representative population pool. The resulting population pool serves as the demographic foundation for the subsequent Behaviour Layer, Opportunity Layer, and mobility simulation framework. The representative population pool is a complete demographic state space generated by integrating all processed demographic datasets without applying behavioural or logical constraints.

All demographic categories are retained, including those with zero counts, to preserve the complete representative state space. Zero counts indicate that the corresponding demographic combination is not supported by the source dataset rather than being removed during preprocessing.

Input datasets:

- Age
- Sex
- Household Composition
- Economic Activity Status
- Industry
- Car Ownership
- Distance Travelled to Work
- Ethnic Group

Output:

- Representative Population Pool

## 1. Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
```

## 2. Set Up File Pathes

```python
from pathlib import Path

project_root = Path.cwd().parents[1]

processed_demographic_path = (
    project_root
    / "data"
    / "demographic"
    / "processed"
)

processed_mobility_path = (
    project_root
    / "data"
    / "mobility"
    / "processed"
)

output_path = (
    project_root
    / "output"
    / "demographic_profiles"
)

output_path.mkdir(parents=True, exist_ok=True)
```

## 3. Load Dataset

```python
# Age
age_file = (
    processed_demographic_path
    / "age"
    / "age.json"
)

# Gender
gender_file = (
    processed_demographic_path
    / "gender"
    / "gender.json"
)

# Household Composition
household_file = (
    processed_demographic_path
    / "household_by_age"
    / "household_by_age.json"
)

# Economic Activity Status
economic_activity_file = (
    processed_demographic_path
    / "economic_activity_status"
    / "economic_activity_status.json"
)

# Industry
industry_file = (
    processed_demographic_path
    / "industry"
    / "industry_by_age.json"
)

# Car Ownership
car_ownership_file = (
    processed_demographic_path
    / "car_ownership_by_household"
    / "car_ownership_by_household.json"
)

# Distance Travelled to Work
travel_distance_file = (
    project_root
    / "data"
    / "mobility"
    / "processed"
    / "travel_distance"
    / "travel_distance_by_age.json"
)

# Ethnic Group
ethnic_group_file = (
    processed_demographic_path
    / "ethnic_group"
    / "ethnic_group_by_age.json"
)
```

```python
age = pd.read_json(age_file)

gender = pd.read_json(gender_file)

household = pd.read_json(household_file)

economic_activity = pd.read_json(economic_activity_file)

industry = pd.read_json(industry_file)

car_ownership = pd.read_json(car_ownership_file)

travel_distance = pd.read_json(travel_distance_file)

ethnic_group = pd.read_json(ethnic_group_file)
```

```python
display(age.head())

display(gender.head())

display(household.head())

display(economic_activity.head())

display(industry.head())

display(car_ownership.head())

display(travel_distance.head())

display(ethnic_group.head())
```

```python
print(age.shape)
print(gender.shape)
print(household.shape)
print(economic_activity.shape)
print(industry.shape)
print(car_ownership.shape)
print(travel_distance.shape)
print(ethnic_group.shape)
```

## 4. Initialise Representative Population Pool

```python
population_pool = age[["Age"]].copy()

display(population_pool)

print(population_pool.shape)
```

## 5. Expand Gender Profiles

```python
population_pool = population_pool.merge(
    gender[["Age","Gender"]],
    on="Age",
    how="left"
)
```

```python
display(population_pool)

print(population_pool.shape)
```

## 6. Expand Household Profiles

```python
population_pool = population_pool.merge(
    household[["Age", "Household Type", "Household Count"]],
    on="Age",
    how="left"
)

display(population_pool.head())

print(population_pool.shape)
```

## 7. Expand Economic Activity Profiles

```python
population_pool = population_pool.merge(
    economic_activity[["Age", "Economic Activity Status", "Economic Count"]],
    on="Age",
    how="left"
)

display(population_pool.head())

print(population_pool.shape)
```

## 8. Expand Industry Profiles

```python
population_pool = population_pool.merge(
    industry[["Age", "Industry", "Industry Count"]],
    on=["Age"],
    how="left"
)
```

```python
display(population_pool.head())

print(population_pool.shape)
```

## 9. Expand Ethnic Group Profiles


The representative population pool is expanded by integrating ethnic group information.

Each representative profile is matched by age group and expanded to include all ethnic group categories supported by the processed dataset.

```python
population_pool = population_pool.merge(
    ethnic_group[["Age", "Ethnic Group", "Ethnic Count"]],
    on="Age",
    how="left"
)

display(population_pool.head())

print(population_pool.shape)
```

## 10. Expand Travel Distance Profiles

```python
population_pool = population_pool.merge(
    travel_distance[
        ["Age", "Distance travelled to work", "Travel Distance Count"]
    ],
    on="Age",
    how="left"
)
```

```python
display(population_pool.head())
print(population_pool.shape)
```

## 11. Expand Car Ownership Profiles


```python
population_pool = population_pool.merge(
    car_ownership,
    on="Household Type",
    how="left"
)

population_pool = population_pool.drop(
    columns=["Lower tier local authorities"]
)
```

```python
display(population_pool.head(15))
print(population_pool.shape)
```

## 12. Validate Representative Population Pool

```python
def validate_dataset(pool, source, keys, count_col, dataset_name):
    source_total = source[count_col].sum()

    pool_total = (
        pool[keys + [count_col]]
        .drop_duplicates()
        [count_col]
        .sum()
    )

    print(f"\n{dataset_name}")
    print(f"Source : {source_total:,}")
    print(f"Pool   : {pool_total:,}")

    if source_total == pool_total:
        print("✅ PASS")
    else:
        print("❌ FAIL")
```

```python
validate_dataset(
    population_pool,
    household,
    ["Age", "Household Type"],
    "Household Count",
    "Household"
)

validate_dataset(
    population_pool,
    economic_activity,
    ["Age", "Economic Activity Status"],
    "Economic Count",
    "Economic"
)

validate_dataset(
    population_pool,
    industry,
    ["Age", "Industry"],
    "Industry Count",
    "Industry"
)

validate_dataset(
    population_pool,
    ethnic_group,
    ["Age", "Ethnic Group"],
    "Ethnic Count",
    "Ethnic"
)

validate_dataset(
    population_pool,
    travel_distance,
    ["Age", "Distance travelled to work"],
    "Travel Distance Count",
    "Travel Distance"
)

validate_dataset(
    population_pool,
    car_ownership,
    ["Household Type", "Car Ownership"],
    "Car Ownership Count",
    "Car Ownership"
)
```

## 13. Export Representative Population Pool

The validated representative population pool was exported after the integration and validation processes were completed.

Two output formats were generated for different purposes:

- **JSON** (`representative_population_pool.json`) was produced as a human-readable representation of the complete representative population pool and for interoperability with other applications.
- **Parquet** (`representative_population_pool.parquet`) was additionally exported as the primary storage format for subsequent framework development because it provides significantly faster read/write performance and more efficient storage for large-scale tabular data.

The representative population pool contains over 21 million representative profiles generated from the combination of demographic attributes and associated mobility-related characteristics. Since the dataset is automatically generated from the processed demographic datasets, it is treated as an output artefact rather than a source file and is therefore excluded from version control.

```python
representative_population_pool = population_pool.copy()

representative_population_pool.to_parquet(
    output_path / "representative_population_pool.parquet",
    index=False,
    compression="zstd"
)
```
