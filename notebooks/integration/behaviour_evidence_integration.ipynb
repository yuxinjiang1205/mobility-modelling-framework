## Overview

This notebook harmonises the 3 selected National Travel Survey (NTS) datasets under a shared trip-purpose classification, and integrate the 3 datasets together to form the Behavioural Evidence Layer used in this framework.

The output shows the mobility representation including trip purpose distribution, trip start time distribution, and travel mode distribution.

Future modellers can add more mobility components into this framework.

## 6. Behaviour Integration

### 6.1 Trip Purpose Mapping

```python
from ontology.nts_purpose_mapping import nts_purpose_mapping
```

### 6.2 NTS0611 Trip-Purpose Distribution Extraction

```python
def extract_nts0611a_trip_purpose_distribution(
    shared_trip_purpose,
):
    nts0611a_trip_purpose_distribution = (
        trip_purpose[
            [
                "Year",
                "Sex",
                "Age",
                shared_trip_purpose,
            ]
        ].copy()
    )

    nts0611a_trip_purpose_distribution.rename(
        columns={
            shared_trip_purpose: (
                "Trip Purpose Probability"
            )
        },
        inplace=True,
    )

    return nts0611a_trip_purpose_distribution
```

```python
nts0611_business = extract_nts0611a_trip_purpose_distribution(
    shared_trip_purpose="Business")

nts0611_business.head()
```

### 6.3 NTS0409a Travel-Mode Distribution Extraction

```python
def extract_nts0409a_travel_mode_distribution(
    shared_trip_purpose,
):
    
    nts0409a_mapped_purposes = (
        nts_purpose_mapping[
            shared_trip_purpose
        ]["nts0409a"]
    )

    nts0409a_travel_mode_distribution = (
        travel_mode[
            travel_mode["Purpose"].isin(
                nts0409a_mapped_purposes
            )
        ].copy()
    )

    return nts0409a_travel_mode_distribution
```

```python
nts0409_business = extract_nts0409a_travel_mode_distribution(
    shared_trip_purpose="Business")

nts0409_business.head()
```

### 6.4 NTS0502b Temporal-Pattern Distribution Extraction

```python
def extract_nts0502b_temporal_distribution(
    shared_trip_purpose,
):
    
    nts0502b_mapped_purposes = (
        nts_purpose_mapping[
            shared_trip_purpose
        ]["nts0502b"]
    )

    nts0502b_trip_start_time_distribution = (
        temporal_pattern[
            temporal_pattern[
                "Trip purpose"
            ].isin(
                nts0502b_mapped_purposes
            )
        ].copy()
    )

    return nts0502b_trip_start_time_distribution
```

```python
nts0502b_business_distribution = (
    extract_nts0502b_temporal_distribution(
        "Business"
    )
)

nts0502b_business_distribution
```

### 6.5 Harmonised NTS Evidence Assembly

```python
def assemble_harmonised_nts_evidence(
    shared_trip_purpose,
):
    
    nts0611a_trip_purpose_distribution = (
        extract_nts0611a_trip_purpose_distribution(
            shared_trip_purpose
        )
    )

    nts0502b_temporal_distribution = (
        extract_nts0502b_temporal_distribution(
            shared_trip_purpose
        )
    )

    nts0409a_travel_mode_distribution = (
        extract_nts0409a_travel_mode_distribution(
            shared_trip_purpose
        )
    )

    return {
        "nts0611a_trip_purpose_distribution": (
            nts0611a_trip_purpose_distribution
        ),
        "nts0502b_temporal_distribution": (
            nts0502b_temporal_distribution
        ),
        "nts0409a_travel_mode_distribution": (
            nts0409a_travel_mode_distribution
        ),
    }
```

```python
business_nts_evidence = (
    assemble_harmonised_nts_evidence(
        "Business"
    )
)

business_nts_evidence.keys()
```

```python
business_nts_evidence[
    "nts0611a_trip_purpose_distribution"
].head()
```

```python
business_nts_evidence[
    "nts0409a_travel_mode_distribution"
]
```

```python
business_nts_evidence[
    "nts0502b_temporal_distribution"
]
```

```python
assembled_nts_evidence = {}

for shared_trip_purpose in nts_purpose_mapping:

    assembled_nts_evidence[
        shared_trip_purpose
    ] = assemble_harmonised_nts_evidence(
        shared_trip_purpose
    )
```

## 8. Export

```python
import json

export_evidence = {}

for shared_trip_purpose, evidence in assembled_nts_evidence.items():

    export_evidence[shared_trip_purpose] = {

        "nts0611a_trip_purpose_distribution": evidence[
            "nts0611a_trip_purpose_distribution"
        ].to_dict(
            orient="records"
        ),

        "nts0502b_temporal_distribution": evidence[
            "nts0502b_temporal_distribution"
        ].to_dict(
            orient="records"
        ),

        "nts0409a_travel_mode_distribution": evidence[
            "nts0409a_travel_mode_distribution"
        ].to_dict(
            orient="records"
        ),
    }
```

```python
Path(output_path).parent.mkdir(
    parents=True,
    exist_ok=True
)

with open(output_path, "w", encoding="utf-8") as f:

    json.dump(
        export_evidence,
        f,
        indent=4,
        ensure_ascii=False,
    )

print(
    "Harmonised NTS evidence exported to:"
    f"\n{output_path}"
)
```
