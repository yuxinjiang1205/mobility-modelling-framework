## Overview

This notebook harmonises and organises behavioural probability evidence derived from three National Travel Survey (NTS) datasets under a unified trip-purpose classification:

- Trip-purpose probabilities by age and gender (NTS0611a)
- Start-time probability distributions by trip purpose (NTS0502b)
- Travel-mode probability distributions by trip purpose (NTS0409a)

As the trip-purpose classifications differ across the three datasets, a predefined mapping is used to link each unified trip purpose to the corresponding categories in NTS0611a, NTS0502b, and NTS0409a.

For each unified trip purpose, the notebook assembles:

- all age- and gender-specific trip-purpose probabilities from NTS0611a;
- one mapped start-time probability distribution from NTS0502b; and
- one mapped travel-mode probability distribution from NTS0409a.

The output is a reusable harmonised behavioural evidence dataset organised by unified trip purpose. It does not sample travel behaviour or construct an individual behavioural representation.

The exported evidence is subsequently queried by the demographic-behavioural integration module, where the relevant NTS0611a probabilities are selected for a representative demographic profile and demographic feasibility constraints are applied. The resulting constrained behavioural representation is then used in the subsequent behaviour-to-POI integration stage.

### 6.2 NTS0611 Trip-Purpose Profile Extraction

```python
def extract_nts0611a_trip_purpose_profiles(
    unified_purpose,
):
    """
    Extract all age- and gender-specific
    trip-purpose probability profiles from NTS0611a
    for one unified trip purpose.
    """
    nts0611a_trip_purpose_profiles = (
        trip_purpose[
            [
                "Year",
                "Gender",
                "Age",
                unified_purpose,
            ]
        ].copy()
    )

    nts0611a_trip_purpose_profiles.rename(
        columns={
            unified_purpose: (
                "Trip Purpose Probability"
            )
        },
        inplace=True,
    )

    return nts0611a_trip_purpose_profiles
```

### 6.3 NTS0409a Travel-Mode Profile Extraction

```python
def extract_nts0409a_travel_mode_profiles(
    unified_purpose,
):
    """
    Extract the travel-mode probability profile(s)
    from NTS0409a mapped to one unified trip purpose.
    """
    mapped_nts0409a_purposes = (
        nts_purpose_mapping[
            unified_purpose
        ]["nts0409a"]
    )

    nts0409a_travel_mode_profiles = (
        travel_mode[
            travel_mode["Purpose"].isin(
                mapped_nts0409a_purposes
            )
        ].copy()
    )

    return nts0409a_travel_mode_profiles
```

### 6.4 NTS0502b Temporal-Pattern Profile Extraction

```python
def extract_nts0502b_temporal_profiles(
    unified_purpose,
):
    """
    Extract the start-time probability profile(s)
    from NTS0502b mapped to one unified trip purpose.
    """
    mapped_nts0502b_purposes = (
        nts_purpose_mapping[
            unified_purpose
        ]["nts0502b"]
    )

    nts0502b_temporal_profiles = (
        temporal_pattern[
            temporal_pattern[
                "Trip purpose"
            ].isin(
                mapped_nts0502b_purposes
            )
        ].copy()
    )

    return nts0502b_temporal_profiles
```

### 6.5 Harmonised NTS Evidence Assembly

```python
def assemble_harmonised_nts_evidence(
    unified_purpose,
):
    """
    Assemble harmonised behavioural evidence from
    NTS0611a, NTS0502b, and NTS0409a for one
    unified trip purpose.

    The assembled evidence contains:
    - all age- and gender-specific trip-purpose
      probability profiles from NTS0611a;
    - the mapped temporal probability profile(s)
      from NTS0502b; and
    - the mapped travel-mode probability profile(s)
      from NTS0409a.
    """
    nts0611a_trip_purpose_profiles = (
        extract_nts0611a_trip_purpose_profiles(
            unified_purpose
        )
    )

    nts0502b_temporal_profiles = (
        extract_nts0502b_temporal_profiles(
            unified_purpose
        )
    )

    nts0409a_travel_mode_profiles = (
        extract_nts0409a_travel_mode_profiles(
            unified_purpose
        )
    )

    return {
        "nts0611a_trip_purpose_profiles": (
            nts0611a_trip_purpose_profiles
        ),
        "nts0502b_temporal_profiles": (
            nts0502b_temporal_profiles
        ),
        "nts0409a_travel_mode_profiles": (
            nts0409a_travel_mode_profiles
        ),
    }
```

```python
business_nts_evidence = (
    assemble_harmonised_nts_evidence(
        "Business"
    )
)

business_nts_evidence.keys()
```

```python
assembled_nts_evidence = {}

for unified_purpose in nts_purpose_mapping:

    assembled_nts_evidence[
        unified_purpose
    ] = assemble_harmonised_nts_evidence(
        unified_purpose
    )
```

## 8. Export

```python
# ==========================================
# Export Harmonised NTS Evidence
# ==========================================

import json

export_evidence = {}

for unified_purpose, evidence in assembled_nts_evidence.items():

    export_evidence[unified_purpose] = {

        "nts0611a_trip_purpose_profiles": evidence[
            "nts0611a_trip_purpose_profiles"
        ].to_dict(
            orient="records"
        ),

        "nts0502b_temporal_profiles": evidence[
            "nts0502b_temporal_profiles"
        ].to_dict(
            orient="records"
        ),

        "nts0409a_travel_mode_profiles": evidence[
            "nts0409a_travel_mode_profiles"
        ].to_dict(
            orient="records"
        ),
    }

output_path = (
    "../../output/behavioural_evidence/"
    "nts_behavioural_evidence_layer.json"
)

with open(output_path, "w", encoding="utf-8") as f:

    json.dump(
        export_evidence,
        f,
        indent=4,
        ensure_ascii=False,
    )

print(
    "Harmonised NTS evidence exported to:"
    f"\n{output_path}"
)
```
