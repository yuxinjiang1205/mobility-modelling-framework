## Overview

This notebook integrates the processed behavioural probability datasets into a unified behaviour representation.

The integration combines three National Travel Survey (NTS) datasets:

- Trip Purpose Probability (NTS0611)
- Temporal Pattern Probability (NTS0502)
- Travel Mode Probability (NTS0409)

As the trip purpose classifications differ across the three datasets, a predefined purpose mapping is applied to harmonise the categories before integration.

For each trip purpose defined in NTS0611, the corresponding temporal pattern and travel mode probability distributions are retrieved from the mapped NTS0502 and NTS0409 datasets.

The output of this notebook is an integrated behavioural probability dataset, which will be used as the behavioural input for the subsequent Opportunity Layer.

The resulting integrated behavioural probability dataset provides a consistent representation of trip purpose, temporal travel patterns, and travel mode distributions. It serves as the behavioural component of the proposed mobility modelling framework and provides the probabilistic inputs required by the Opportunity Layer.

### Step 6.2 Baseline Profile Extraction

```python
def extract_baseline_profile(unified_purpose):
    """
    Extract the baseline trip purpose profile from NTS0611.
    """

    profile = trip_purpose[
        ["Year", "Sex [note 2]", "Age", unified_purpose]
    ].copy()

    profile.rename(
        columns={
            unified_purpose: "Trip Purpose Probability"
        },
        inplace=True
    )

    return profile
```

### 6.3 Travel Mode Mapping

```python
def extract_travel_mode(unified_purpose):
    """
    Extract travel mode probabilities from NTS0409
    using the trip purpose mapping.
    """

    mapped_purposes = purpose_mapping[unified_purpose]["nts0409"]

    profile = travel_mode[
        travel_mode["Purpose"].isin(mapped_purposes)
    ].copy()

    return profile
```

```python
extract_travel_mode("Business")
```

```python
extract_travel_mode("Personal Business")
```

### 6.4 Temporal Pattern Mapping

```python
def extract_temporal_pattern(unified_purpose):
    """
    Extract temporal pattern probabilities from NTS0502
    using the trip purpose mapping.
    """

    mapped_purposes = purpose_mapping[unified_purpose]["nts0502"]

    profile = temporal_pattern[
        temporal_pattern["Trip purpose"].isin(mapped_purposes)
    ].copy()

    return profile
```

```python
extract_temporal_pattern("Business")
```

```python
extract_temporal_pattern("Holiday: Base")
```

### 6.5 Behaviour Profile Construction

```python
def construct_behaviour_profile(unified_purpose):
    """
    Construct the integrated behavioural profile
    for one unified trip purpose.

    The profile combines:
    - Baseline trip purpose probability (NTS0611)
    - Travel mode probability (NTS0409)
    - Temporal pattern probability (NTS0502)
    """

    baseline_profile = extract_baseline_profile(unified_purpose)

    travel_mode_profile = extract_travel_mode(unified_purpose)

    temporal_pattern_profile = extract_temporal_pattern(unified_purpose)

    return {
        "baseline_profile": baseline_profile,
        "travel_mode_profile": travel_mode_profile,
        "temporal_pattern_profile": temporal_pattern_profile
    }
```

```python
all_behaviour_profiles = {}

for purpose in purpose_mapping:

    all_behaviour_profiles[purpose] = construct_behaviour_profile(purpose)
```

## 8. Export

```python
# ==========================================
# Export Behaviour Profiles
# ==========================================

import json

export_profiles = {}

for purpose, profile in all_behaviour_profiles.items():

    export_profiles[purpose] = {

        "baseline_profile": profile["baseline_profile"].to_dict(
            orient="records"
        ),

        "travel_mode_profile": profile["travel_mode_profile"].to_dict(
            orient="records"
        ),

        "temporal_pattern_profile": profile["temporal_pattern_profile"].to_dict(
            orient="records"
        )

    }

output_path = "../../outputs/behaviour_profiles.json"

with open(output_path, "w", encoding="utf-8") as f:

    json.dump(export_profiles, f, indent=4)

print(f"Behaviour profiles exported to:\n{output_path}")
```
