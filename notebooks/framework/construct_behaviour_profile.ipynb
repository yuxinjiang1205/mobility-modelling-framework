# Construct Behaviour Profile

## Overview

This notebook constructs a representative behaviour profile for an individual representative demographic profile within the proposed human mobility representation framework.

Unlike the previous integration stage, which constructs the Representative Population Pool, this notebook focuses on linking demographic characteristics with mobility-related behavioural distributions derived from the National Travel Survey (NTS).

Given a representative demographic profile, the framework retrieves the corresponding behavioural distributions based on the processed mobility datasets, including:

- Trip purpose distribution
- Temporal distribution
- Travel mode distribution

Rather than generating a single realised journey through sampling, this notebook preserves the complete behavioural representation of an individual by retaining all statistically plausible behavioural possibilities and their associated probabilities.

The resulting behaviour profile serves as the behavioural representation of one representative person and forms the foundation for the subsequent Opportunity Layer within the proposed framework.

---

## Objectives

The objectives of this notebook are:

- Load the processed mobility behaviour datasets.
- Accept one representative demographic profile as input.
- Retrieve the corresponding trip purpose distribution.
- Associate each trip purpose with its temporal distribution.
- Associate each trip purpose with its travel mode distribution.
- Construct a complete representative behaviour profile without performing behavioural sampling.
- Export the behaviour profile for subsequent framework integration.

> **Note**

This notebook is designed to construct behavioural representations rather than simulate realised human movements. Behavioural sampling is considered a downstream application of the proposed framework and is therefore outside the scope of this notebook.

## Step 1 — Load Processed Datasets

```python
from pathlib import Path
import json
import pandas as pd

project_root = Path.cwd().parents[1]

population_pool_file = (
    project_root
    / "output"
    / "demographic_profiles"
    / "representative_population_pool.parquet"
)

behaviour_profiles_file = (
    project_root
    / "output"
    / "behaviour_profiles"
    / "behaviour_profiles.json"
)
```

```python
results = {}

for purpose in behaviour_profiles.keys():

    baseline = behaviour_profiles[purpose]["baseline_profile"]

    match = [
        row for row in baseline
        if row["Age"] == person_age
        and row["Sex"] == person_sex
    ]

    results[purpose] = match[0]
```

```python
results
```

```python
results = {}

for purpose, evidence in behaviour_profiles.items():

    baseline = evidence["baseline_profile"]

    match = [
        row for row in baseline
        if row["Age"] == person_age
        and row["Sex"] == person_sex
    ]

    if not match:
        raise ValueError(
            f"No baseline profile found for "
            f"purpose={purpose}, age={person_age}, sex={person_sex}"
        )

    results[purpose] = {
        "trip_purpose_probability": match[0]["Trip Purpose Probability"],
        "travel_mode_profile": evidence["travel_mode_profile"][0],
        "temporal_pattern_profile": evidence["temporal_pattern_profile"][0]
    }
```
